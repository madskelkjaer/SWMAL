#### Resume: Look at the Big Picture

In the end-to-end pipeline of Machine learning, it is important to ”zoom out”
and look at the bigger picture. What do we want to achieve with this model?
Ideally, we would start by defining a problem and then see which specific data
we need, and additionally get an idea of how we would process them to reach
an end goal.

#### Resume: Get the Data

To answer the before mentioned questions, we must gather some data. First,
we must decide on the method we want to use to gather the data. Afterwards, ideally very early in the process, we can start splitting the data. We split it into test- and training-data to teach our model how to generalize the data.

#### Resume: Explore and Visualize the Data to Gain Insights,

Once the data is collected, the next step is to explore and visualize it. This helps us gain a better understanding of its structure, quality, and possible patterns.
Visualization and simple statistics allow us to spot errors, biases, or anomalies in the data. As the saying goes, ”bad data in, bad data out” — so it is crucial to ensure that the data makes sense before continuing.


#### Resume: Prepare the Data for Machine Learning Algorithms

Raw data is rarely ready to be used directly. Preparation may include clean-
ing missing values (for example, by replacing them with the mean or median),
normalizing or scaling numerical values, and transforming categorical variables into numerical ones (e.g. one-hot encoding). At this stage, we also ensure the data format matches what our machine learning algorithms expect.


#### Resume: Select and Train a Model

With the data prepared, we can select an appropriate model. This often begins with simpler algorithms to establish a baseline and then progresses to more advanced techniques. The model is trained on the training data to capture the patterns necessary to solve the defined problem.

#### Resume: Fine-Tune Your Model

After training, the model must be evaluated on previously unseen test data. This evaluation provides insight into how well the model generalizes. Depending on the problem type, metrics such as accuracy, precision, recall, F1-score, or RMSE are used. A key challenge is avoiding overfitting, where the model performs well on training data but poorly on new data.

If performance is not satisfactory, the model can be improved by tuning hyperparameters or trying alternative algorithms. Techniques like grid search, random search, or cross-validation can systematically optimize performance. Feature selection or engineering may also provide gains.

#### Resume: Launch, Monitor, and Maintain Your System

Once the model is performing well, the next step is deployment. This means
integrating the model into a real-world application, such as through an API, a dashboard, or by embedding it in an existing system where it can generate predictions on new data.
A model’s job does not end once it is deployed. Over time, data distributions may shift, leading to performance degradation (concept drift). To ensure continued reliability, the model must be monitored in production, retrained periodically, and updated to handle new data or requirements.

### a. Concept

We want to build an image-classifier that predicts the cat breed from a photo. We start with five classes: Bengal, Domestic Shorthair, Maine Coon, Ragdoll, Siamese. But we wish to keep the pipeline extensible so we can add more breeds later with minimal changes.

### b. Dataset

We use the [Kaggle Cat's Breed Dataset](https://www.kaggle.com/datasets/yapwh1208/cats-breed-dataset) which contains the five folders listed above. The directory-per-breed layout fits standard loaders. It would be possible to find other datasets with more cat breeds or just with more pictures for each race by adding their folders to the split and rerunning the pipeline.

### c. Dataset description

Each sample is a color image (mostly RGB, JPEG/PNG) with varying sizes. Labels are the folder names. From our exploration we know the directory layout, class counts, typical sizes/aspects, and have quick visual checks via example grids and single-image RGB histograms. Potential issues: minor label noise, class imbalance, varied backgrounds, and resolution variance.

### d. Use of dataset

We’ll use the dataset for multiclass classification: given a cat photo, predict one breed among our five classes. Classification fits because the target is a discrete label, not a numeric value (so not regression). We’ll resize and convert to RGB to standardize inputs and stabilize training, then make stratified train/validation/test splits to get fair evaluation. For metrics, we pick accuracy for overall performance and macro-F1 so small classes count equally (We might add a cat breed with not a lot of pictures). We will use a confusion matrix that shows us which breeds the model confuses.

We start by setting the dataset path, importing the libraries, and configuring plotting so the rest of the notebook knows where the images are and can visualize results consistently.

```python
# Setup: Datapath and imports
DATA_DIR = "CatData/cat_v1"

from pathlib import Path
import numpy as np
import pandas as pd
from PIL import Image, UnidentifiedImageError
import matplotlib.pyplot as plt

plt.rcParams["figure.figsize"] = (8,5)
plt.rcParams["axes.grid"] = True
```

Next, we scan the class folders, open each image, and record format, color mode, width, height, and aspect ratio to build a simple metadata table.

```python
# Scan dataset (formats, color spaces, sizes)
IMG_EXTS = {".jpg",".jpeg",".png",".bmp",".gif",".tif",".tiff",".webp"}

rows, bad = [], []
root = Path(DATA_DIR)
classes = sorted([d.name for d in root.iterdir() if d.is_dir()])

for cls in classes:
    for p in (root/cls).rglob("*"):
        if p.is_file() and p.suffix.lower() in IMG_EXTS:
            try:
                with Image.open(p) as im:
                    im.load()
                    w, h = im.size
                    rows.append({
                        "path": str(p),
                        "label": cls,
                        "format": im.format,    # JPEG, PNG, ...
                        "mode": im.mode,        # RGB, L, RGBA, ...
                        "width": w,
                        "height": h,
                        "aspect": (w/h if h else np.nan)
                    })
            except (UnidentifiedImageError, OSError):
                bad.append(str(p))

meta = pd.DataFrame(rows)
print(f"Classes: {classes}")
print(f"Total images: {len(meta)}  |  Unreadable: {len(bad)}")
meta.head()
```

Then we print a directory tree to document how the data are organized on disk, showing the root and the class subfolders.

```python
# Representation on disk (folder structure)
def print_tree(path: Path, max_depth=1):
    def _walk(p, prefix="", depth=0):
        if depth > max_depth: return
        entries = sorted(p.iterdir(), key=lambda x: (not x.is_dir(), x.name.lower()))
        for i, e in enumerate(entries[:50]):  # truncate for readability
            is_last = (i == len(entries)-1)
            branch = "└── " if is_last else "├── "
            print(prefix + branch + e.name + ("/" if e.is_dir() else ""))
            if e.is_dir():
                _walk(e, prefix + ("    " if is_last else "│   "), depth+1)
    print(p.name + "/")
    _walk(path)

print_tree(root, max_depth=1)
```

After that, we show class counts and simple tables of image formats and color modes. This checks how balanced the classes are and what kinds of files we have.

```python
# Basic tables
# Class counts
cls_counts = meta["label"].value_counts().sort_index()
display(pd.DataFrame({"count": cls_counts}))

# Formats & color spaces
display(meta["format"].value_counts().to_frame("count").rename_axis("format"))
display(meta["mode"].value_counts().to_frame("count").rename_axis("mode"))
```

We then compute size stats per class and plot histograms of width, height, and aspect ratio. This tells us how big images are and helps if we want to choose a standard resize.

```python
# Size statistics + histograms (width/height/aspect)
size_stats = (meta
    .assign(megapix=(meta["width"]*meta["height"])/1e6)
    .groupby("label")[["width","height","aspect","megapix"]]
    .agg(["count","min","median","mean","max"]))
display(size_stats)

fig, axes = plt.subplots(1,3, figsize=(14,4))
axes[0].hist(meta["width"],  bins=40); axes[0].set_title("Width (px)");  axes[0].set_xlabel("px")
axes[1].hist(meta["height"], bins=40); axes[1].set_title("Height (px)"); axes[1].set_xlabel("px")
axes[2].hist(meta["aspect"], bins=40); axes[2].set_title("Aspect ratio (w/h)"); axes[2].set_xlabel("ratio")
plt.tight_layout(); plt.show()
```

Next, we show a small grid of example images for each class with their labels. This is a quick visual check that the data look right.

```python
# Examples & labels (image grid)
def show_examples(df, per_class=4):
    labels = sorted(df["label"].unique())
    nrows, ncols = len(labels), per_class
    fig, axes = plt.subplots(nrows, ncols, figsize=(2.8*ncols, 2.8*nrows))
    if nrows == 1: axes = np.array([axes])
    for r, lab in enumerate(labels):
        subset = df[df["label"]==lab]
        sample = subset.sample(min(per_class, len(subset)), random_state=42)
        for c, (_, row) in enumerate(sample.iterrows()):
            ax = axes[r, c] if ncols>1 else axes[r, 0]
            img = Image.open(row["path"]).convert("RGB")
            ax.imshow(img); ax.set_axis_off()
            if c == 0: ax.set_title(lab)
    plt.suptitle("Example images per class", y=0.995)
    plt.tight_layout(); plt.show()

show_examples(meta, per_class=4)
```

We also plot RGB histograms for one image per class. That gives a basic view of brightness and color for single images.

```python
# Histograms on individual images
def plot_image_and_histogram(image_path, bins=32):
    im = Image.open(image_path).convert("RGB")
    arr = np.asarray(im, dtype=np.uint8)
    fig = plt.figure(figsize=(10,4))
    ax1 = plt.subplot(1,2,1)
    ax1.imshow(im); ax1.set_axis_off(); ax1.set_title(Path(image_path).parent.name)
    ax2 = plt.subplot(1,2,2)
    for ch, name in enumerate(["R","G","B"]):
        hist, edges = np.histogram(arr[...,ch].ravel(), bins=bins, range=(0,255))
        centers = 0.5*(edges[1:]+edges[:-1])
        ax2.plot(centers, hist, label=name)
    ax2.set_title("Pixel-value histogram"); ax2.set_xlabel("0–255"); ax2.legend()
    plt.tight_layout(); plt.show()

print("Single-image histograms (one per class):")
for lab in sorted(meta["label"].unique()):
    ex = meta[meta["label"]==lab]
    if len(ex):
        plot_image_and_histogram(ex.iloc[0]["path"], bins=32)
```
